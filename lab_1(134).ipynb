```python
import numpy as np

np.random.seed(42)
marks = np.array([55, 72, 48, 90, 66, 81])

print('Marks:', marks)
print('Mean:', marks.mean())
print('Highest:', marks.max())
print('Passed:', marks[marks >= 50])


print('Minimum:', marks.min())
print('Standard Deviation:', marks.std())

bonus_marks = marks + 5
print('Bonus Marks:', bonus_marks)
print('Original Marks:', marks)

passed_count = np.sum(marks >= 50)
print('Number of Passed Students:', passed_count)

matrix = marks.reshape(2, 3)
print('2 × 3 Matrix:')
print(matrix)
```

```python
import pandas as pd

data = {
    'student': ['Asha', 'Babu', 'Chayan', 'Dipa', 'Emon', 'Farah'],
    'study_hours': [2, 5, 1, 7, 3, 6],
    'sleep_hours': [8, 6, 9, 5, 7, 6],
    'attendance': [70, 88, 55, 95, 75, 91],
    'snacks': [3, 2, 5, 1, 4, 2],
    'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

display(df)

print(df.info())

print(df.describe())


# 16. Features and Label
print("Features:", ['study_hours', 'sleep_hours', 'attendance', 'snacks'])
print("Label:", 'passed')


# 17. Samples and Columns
print("Samples:", df.shape[0])
print("Columns:", df.shape[1])


# 18. Attendance at least 80
print("Attendance >= 80:")
display(df[df['attendance'] >= 80])


# 19. Average study hours of passed students
print("Average study hours:",
      df[df['passed'] == 'Yes']['study_hours'].mean())


# 20. Sort by attendance descending
print("Sorted by attendance:")
display(df.sort_values('attendance', ascending=False))


# 21. Explanation
print("No, this tiny dataset does not prove that snack purchases cause failure.")
print("The dataset has only 6 students, and other factors may affect the result.")
print("Correlation does not necessarily mean causation.")
```
